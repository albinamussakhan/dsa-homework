```python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
```

```python
#База
#1
cities_clean=[city.strip().capitalize() for city in cities]
print(cities_clean)
#2
amounts_clean=[int(amount) for amount in amounts]
print(amounts_clean)
#3
count=0
sum=0
for i in range(len(amounts_clean)):
 if paid[i]==True:
  count+=1
  sum+=amounts_clean[i]
ave=sum/count
print(f'Total revenue: {sum}, Average: {ave}, Orders: {count}')
#4
unpaid=0
sum2=0
for i in range(len(amounts_clean)):
  if paid[i]==False:
    unpaid+=1
    sum2+=amounts_clean[i]
for city, amount, status in zip(cities_clean, amounts_clean, paid):
  print(f' {city} | {amount} | {status}')

```

```python
#Стандарт
#5
ala=0
ast=0
for city, status, amount in zip(cities_clean, paid, amounts_clean):
  if status==True and city=='Astana':
    ast+=amount
  if status==True and city=='Almaty':
    ala+=amount
print(f'Astana: {ast} Almaty: {ala}')
#6
max_amount = 0
max_city = ""
max_index = 0

for i, (city, amount, status) in enumerate(zip(cities_clean, amounts_clean, paid)):
    if status == True and amount > max_amount:
        max_amount = amount
        max_city = city
        max_index = i

print(f"Max: {max_amount}, City: {max_city}, Index: {max_index}")
#7
weekend=0
for day in days:
  if day=='сб' or day=='вс':
    weekend+=1
print(f'Weekend: {weekend/len(days)*100:.2f}%')
#9
#case1
budget=15000
course=400000
savings=0
months=0
while savings<course:
  savings+=budget
  months+=1
print(f'It will take {months} months!')
#case2
budget=15000
course=400000
savings=0
months=0
while savings<course:
  savings+=budget
  budget+=1000
  months+=1
print(f'It will take {months} months!')
```

```python
#10
unique_cities=[]
rev_by_city=[]
sorted_rev=[]
sorted_cities=[]

for city,status,amount in zip(cities_clean,paid,amounts_clean):
    if city not in unique_cities:
        unique_cities.append(city)

for ucity in unique_cities:
    rev_city=0
    for city,amount,status in zip(cities_clean,amounts_clean,paid):
        if city==ucity and status==True:
            rev_city+=amount
    rev_by_city.append(rev_city)

sorted_rev=sorted(rev_by_city,reverse=True)

for revenue in sorted_rev:
    for i in range(len(rev_by_city)):
        if rev_by_city[i]==revenue and unique_cities[i] not in sorted_cities:
            sorted_cities.append(unique_cities[i])
            break

unique_cities=sorted_cities
revenue_by_city=sorted_rev

print(unique_cities)
print(revenue_by_city)

#11
sorted_amounts=sorted(amounts_clean)
n=len(sorted_amounts)

if n%2==1:
    median=sorted_amounts[n//2]
else:
    middle1=sorted_amounts[n//2-1]
    middle2=sorted_amounts[n//2]
    median=(middle1+middle2)/2

print("Median:",median)

#12
days_clean=[]

for day in days:
    days_clean.append(day.strip().lower())
unique_days=[]
revenue_by_day=[]

for day in days_clean:
    if day not in unique_days:
        unique_days.append(day)

for uday in unique_days:
    rev_day=0
    for day,amount,status in zip(days_clean,amounts_clean,paid):
        if day==uday and status==True:
            rev_day+=amount
    revenue_by_day.append(rev_day)

for i in range(len(unique_days)):
    blocks=revenue_by_day[i]//5000
    print(f"{unique_days[i]} | {'█'*blocks}  {revenue_by_day[i]}")
```
